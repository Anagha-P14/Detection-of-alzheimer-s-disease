```python
#importing all the libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_validate
from sklearn.model_selection import cross_val_score, cross_val_predict
from sklearn import metrics
from sklearn.model_selection import KFold
from sklearn.model_selection import StratifiedKFold
from sklearn.decomposition import PCA
from scipy import linalg, mat, dot
from sklearn.linear_model import LogisticRegression
from sklearn import preprocessing
import matplotlib.pyplot as plt
from scipy import interp
from sklearn.datasets import make_classification
```

```python
seed = 8
np.random.seed(seed)
#Importing the input csv file
data = pd.read_csv("DWT_wavelet.csv" )
data.drop(data.columns[[0]], axis=1, inplace=True)
data.drop(data.head(1).index, inplace=True)
#create data frame into numpy array

dataset=data.to_numpy()
print(dataset.shape)
X = dataset [:, :-1]
Y = dataset[:,-1]
```

```python
X=np.array(X)
print (X.shape)
Y=np.array(Y)
print(X.shape)
print(Y.shape)
#Provides train/test indices to split data in train/test sets.
kf = StratifiedKFold(n_splits=10, shuffle=True, random_state=seed)
#importing the logistic regresion model from Sklearn library
log_class = LogisticRegression()
# Divide the data into four points and calculate the result of each cross-validation and return
print('cross validation score')
a = cross_val_score(log_class, X, Y, scoring='accuracy', cv = kf)
print(a)
print('cross validation prediction')
y_pred = cross_val_predict(log_class, X, Y, cv=kf)
print(y_pred)
conf_mat = confusion_matrix(Y, y_pred)
```

```python
sum = 0
for k in a:
 sum = sum + k
 avg = 100 * (sum/10)
print('a:')
print(a)
print('avg:')
print(avg)
print(' Confusion matrix:')
print(conf_mat)
TP = conf_mat[1, 1]
TN = conf_mat[0, 0]
FP = conf_mat[0, 1]
FN = conf_mat[1, 0]
classification_accuracy = (TP + TN) / float(TP + TN + FP + FN)
false_positive_rate = FP / float(TN + FP)
precision = TP / float(TP + FP)
print(' Prediction Accuracy :')
print(classification_accuracy)
```
